```python
def outer_function():
    print("Inside Outer Function")
    enclosing_var = "Enclosing Variable"

    def inner_function():
        local_var = "Local Variable"
        print("Inside inner_function:")
        print("Local Variable:",local_var)
        print("Enclosing Variable:",enclosing_var)

    inner_function()

outer_function()
```

```python
def outer_function():
    print("Inside Outer Function")
    enclosing_var = "Enclosing Variable"

    def inner_function():
        local_var = "Local Variable"
        print("Inside inner_function:")
        print("Local Variable:",local_var)
        print("Enclosing Variable:",enclosing_var)

    inner_function()
    print("Local Variable:", local_var)

outer_function()
```

```python
def outerFunc():
    sample_str = 'Outer Function'
    def innerFunc():
        sample_str = 'Inner Function'
        print(sample_str)
    innerFunc()
    print(sample_str)
outerFunc()
```

```python
#Fibonacci
def Fibonacci(n):
    i = 1
    while(i<=n):
        n=(n-1)+(n+2)
        print(n)
n=int(input("Enter the nuber"))
print("Fibonacci of the number is :",Fibonacci(n))
```
