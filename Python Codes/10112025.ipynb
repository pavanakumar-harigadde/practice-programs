```python
file_path = 'C:/PES/Python Programming/PP/new_directory'
file_stats = os.stat(file_path)
print("File Size in Bytes :",file_stats.st_size)
print("File Size in Kilobytes :",file_stats.st_size/1024)
print("File Size in Megabytes :",file_stats.st_size/(1024*1024))
print("File Creation Time :",file_stats.st_ctime)
print("File Modification Time :",file_stats.st_mtime)
print("File Access Time :",file_stats.st_atime)
```
