```python
num= [10,20,30,40,50]

#append method
num.append(60)
print(num)

#pop method
num.pop()
print(num)

#remove method
num.remove(50)
print(num)

#reverse method
num.reverse()
print(num)

print(num[1])
print(num[-1])

#sort method
num.sort()
print(num)



```
