```python
words=['apple','cat','orange','egg','ball']

result = list(
    map(lambda w:w[0],
        filter(lambda w:w[0].lower() in 'aeiou',words)
       ))
print(result)
```

```python
from functools import reduce
nums = [1,2,3,4]
result = reduce(lambda acc,x:acc + x,nums)
print(result)
```

```python
import functools
pairs=[(1,'a'),(2,'b'),(3,'c')]
functools.reduce(lambda acc, pair: acc+pair[0],pairs,0)
```

```python
def sum_first_elements(pairs):
    acc=0
    for pair in pairs:
        acc+=pair[0]
    return acc
    
pairs=[(1,'a'),(2,'b'),(3,'c')]
print(sum_first_elements(pairs))
```

```python
nums=[1,2,3,4,5,6]

result = reduce(
    lambda acc, x:acc+x,
    map(lambda x:x*x,
        filter(lambda x:x%2==0,nums))
    ,0)

print(result)
```

```python
from itertools import zip_longest
numbers = [1,2,3]
letters = ['a','b','c']
longest = range(5)
zipped = zip_longest(numbers, letters, longest, fillvalue="?")
list(zipped)
```
