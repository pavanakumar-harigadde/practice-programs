```python
class Mobile:
    device_category="Smartphone"
    
    def __init__(self, brand='Samsung', model='S25', ram=12, storage=512):
        self.brand=brand
        self.model=model
        self.ram=ram
        self.storage=storage

    def __new__(cls, *args,**kwargs):
        print("Object is being created now!")
        obj=super().__new__(cls)
        return obj

    def display_details(self):
        print("Mobile Details: ")
        print("Device Category: ", Mobile.device_category)
        print(f"Brand:{self.brand}")
        print(f"Model: {self.model}")
        print(f"RAM: {self.ram} GB")
        print(f"SSD: {self.storage} GB")

    def update_category(cls, new_category):
        print(f"Current category: {Mobile.device_category}")
        Mobile.device_category=new_category
        print(f"The category is changed to {Mobile.device_category}") 

    def is_valid_storage(self):
        if self.storage>0:
            return True
        else:
            return False
```

```python
m=Mobile()
m.display_details()
```

```python
n=Mobile('Apple','iPhone', 8,256)
n.display_details()
```

```python
n.is_valid_storage()
```
