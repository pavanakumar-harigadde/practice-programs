```python
import csv
with open('students.csv', 'r') as file:
    reader = csv.DictReader(file)
summary = []
for row in reader:
    avg = float(row['Marks'])
    result = "Pass" if avg >= 40 else "Fail"
    summary.append({'RollNo': row['RollNo'], 'Name': row['Name'], 'AverageMarks':avg, 'Result': result})
with open('student_summary.csv', 'w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=['RollNo', 'Name', 'AverageMarks', 'Result'])
    writer.writeheader()
    writer.writerows(summary)
```

```python
import csv
sales = {}
with open('daily_sales.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        sales[row['Region']] = sales.get(row['Region'], 0) + float(row['Amount'])
with open('region_sales.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Region', 'TotalSales'])
    for reg, total in sales.items():
        if total >= 50000: # Exclude below 50000
            writer.writerow([reg, total])
```
