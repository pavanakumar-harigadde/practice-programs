```python
class laptop:

    def __init__(self,brand, model,price):
        self.brand = brand
        self.model = model
        self.price = price
        self.battery_level = 100

    def turn_on(self):
        print(f"{self.brand} {self.model} is now ON")

    def turn_off(self):
        print(f"{self.brand} {self.model}  is now OFF")

    def display_info(self):
        print(f"Laptop Details : \n Brand: {self.brand} \n Model:{self.model} \n Price: {self.price} ")

    def check_battery(self):
        print(f"This laptop has {self.battery_level} % battery remaining.")

    def update_battery_level(self, new_level):
        if 0 <= new_level <= 100:
            self.battery_level = new_level
            print(f"Battery level updated to {self.battery_level}%")
        else:
            print("Invalid battery level! Please enter a value between 0 and 100.")

    def decrease_battery(self, usage):
        if usage < 0:
            print("Usage amount cannot be negative!")
        elif self.battery_level - usage <0 :
            print("Battery too low! Please charge your laptop.")
            self.battery_level = 0
        else :
            self.battery_level -= usage
            print(f"Battery decreased by {usage}%. Current level : {self.battery_level}%.")
```

```python
laptop1 = laptop("Dell", "Inspiron 15", 55000,100)
print(laptop1.turn_on())
print(laptop1.turn_off())
print(laptop1.display_info())
print(laptop1.check_battery())
print(laptop1.update_battery_level(85))
```

```python
laptop2 = laptop("Apple", "Macbook Air", 120000, 100)
print(laptop2.turn_on())
print(laptop2.turn_off())
print(laptop2.display_info())
print(laptop2.check_battery())
print(laptop1.update_battery_level(50))
```

```python
class Bank_Account:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder=account_holder
        self.__account_number=account_number
        self.__balance=balance

    def show(self):
        print("Account Holder:",self.account_holder)
        print("Account Number:",self.__account_number)
        print("Account Balance:",self.__balance)

    def deposit(self,amount):
        if(amount<=0):
            print("Please enter the valid amount to deposit.")

        else:
            self.__balance=self.__balance+amount
            print("The amount of ",amount,"is deposited to the account")
            print("Current balance in the account: ", self.__balance)

    def withdraw(self,amount):
        if(amount>self.__balance):
            print("Insufficient Balance.")
        if(amount<=0):
            print("Please enter valid amount to withdraw.")
        else:
            self.__balance=self.__balance-amount
            print("The amount of ",amount,"is  withdrawn from the account")
            print("Current Balance :", self.__balance)

    
    def get_balance(self):
        print("Current balance in the account :",self.__balance)
    
```

```python
c=Bank_Account('Mark', 1234,1000)
c.show()
```

```python
c.deposit(100)
```

```python
c.withdraw(100)
```

```python
c.get_balance()
```
