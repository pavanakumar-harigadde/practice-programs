```python
class Employee:
    company_name="Tech Solutions"
    total_employees=0

    def __init__(self, name, emp_id, salary):
        self.name=name
        self.emp_id=emp_id
        self.salary=salary
        Employee.total_employees= Employee.total_employees+1

    def display_emp_info(self):
        print("\nEmployee Details:")
        print(f"Employee Name : {self.name}")
        print(f"Emplyee ID: {self.emp_id}")
        print(f"Salary: {self.salary}")
        print("Total Employees: ", Employee.total_employees)
        print("\n")

    def update_company(cls, new_company):
        print("Previous Company: ", Employee.company_name)
        Employee.company_name=new_company
        print("New Company : ", Employee.company_name)
```

```python
e1=Employee('Mukesh',1,30000)
e2=Employee('Gukesh',2,100000)
e3=Employee('Yogesh',3,1000)
e1.display_emp_info()
e2.display_emp_info()
e3.display_emp_info()
```
