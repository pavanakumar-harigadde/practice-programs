```python
# in amount column fill missing values with mean or median
df_sales["Amount"] = df["Amount"].fillna(df_sales["Amount"].median())
```

```python
df_sales
```

```python
df_sales["Amount"] = df["Amount"].fillna(df_sales["Amount"].mean())
```

```python
df_sales
```

```python
#To drop a column 
df_sales = df_sales.drop(columns=['City'])
```

```python
df_sales = df_sales.dropna(axis=1,how='all')
df_sales
```

```python
rename_map = {"Prod":"Products","Qty":"Quantity"}
df_sales = df_sales.rename(columns = rename_map)
df_sales.columns
```

```python
high_sales = df_sales['Amount'] > 2000
high_sales
```

```python
df_high = df_sales[high_sales]
len(df_high)
```

```python
df_high
```

```python
#combine the multiple conditions with & and |
cond_amount = df['Amount'] > 2000
cond_region = df['Region'] == "south"
```

```python
#To extract the product detail with amount > 2000 from south
df_target = df_sales[cond_amount & cond_region]
df_target
```

```python
df_target = df_sales[cond_amount | cond_region]
df_target
```

```python
df_sorted = df_sales.sort_values("Amount", ascending=False)
df_sorted
```

```python
df_sorted = df_sales.sort_values(["Region","Amount"])
df_sorted
```

```python
#creating the derived column using algorithmic opeartion
#creating the "Tax" column by 10% of Amount
df_sales["Tax"] = df_sales["Amount"] * 0.10
```

```python
df_sales
```

```python
df_sales["TotalWithTax" ] = df_sales['Amount'] + df_sales["Tax"]
df_sales
```

```python
#To count the occurance of values (in categorica)
df_sales = df_sales["Region"].value_counts()
df_sales
```
