```python
#create a table
cur = conn.cursor()
conn.rollback()
cur.execute(
    """
        CREATE TABLE students (students_id SERIAL PRIMARY KEY,
        student_name VARCHAR(10) NOT NULL,
        grade VARCHAR(10) NOT NULL,
        section VARCHAR(5) NOT NULL 
        );
    """
)

conn.commit()
cur.close()
conn.close()
```
