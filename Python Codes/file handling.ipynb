```python
import csv
with open("C://PES/Python Programming/file1.csv", 'r') as file_obj2 :
    reader = csv.reader(file_obj2)
    line_count=0
    for row in reader:
        if line_count==0:
            print(f'Column names are {",".join(row)}')
            line_count+=1
        else:
            print(f'\t{row[0]} age is {row[1]} and profession is {row[2]}.')
            line_count+=1
    print(f"Processed {line_count} lines.")
```

```python
import csv
data = [
    ['Name', 'Age', 'City'],
    ['Abbott',25,'New York'],
    ['Bob',30,'Los Angeles'],
    ['Chang',35,'Chicago'],
    ['Amar',25,'New York'],
    ['Biswas',30,'San Fransisco'],
    ['Chay',35,'Los Angeles'],
    ['Dhruv',29,'Chicago'],
    ['Eshan',28,'New York'],
    ['Farhan',23,'California'],
    ['Gauging',21,'Las Vegas'],
]

with open('C://PES/Python Programming/output.csv', 'w+', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)
    file.seek(0)
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
