```python
#define a list
my_list = [4,7,0,3]
#get an iterator using iter()
my_iter = iter(my_list)
 
print(my_list)
print(my_iter)
## iterate through it using next()
print(next(my_iter))
print(next(my_iter))
## next(obj) is same as obj.__next__()
print(my_iter.__next__())
print(my_iter.__next__())
## This will raise error , no items left
next(my_iter)
print("")
```

```python
my_list=[1,2,3,4,5]
my_iterator=iter(my_list) #Create an iterator from a list
#Iterate through the list using the iterator
try:
    while True:
        element=next(my_iterator)
        print(element)
except StopIteration:
    pass
```

```python
def make_pretty(func):
    #
    def inner():
        print("I got decored!")
        #
        func()
    #
    return inner
```

```python
def ordinary():
    print("I am ordinary")
```

```python
decorated_func=make_pretty(ordinary)
```

```python
decorated_func()
```

```python
def outer_div(func):
    def inner(x,y):
        if(x<y):
            x,y=y,x
            return func(x,y)
    return inner
```

```python
@outer_div
def divide(x,y):
    print(x/y)
```

```python
divide(2,8)
```

```python
def split_string(function):
    def wrapper():
        func=function()
        splitted_string=func.split()
        return splitted_string

    return wrapper
```

```python
def say_hi():
    return 'hello there'
```

```python
def uppercase_decorator(function):
    def wrapper():
        func=function()
        make_uppercase=func.upper()
        return make_uppercase

    return wrapper
```

```python
@split_string
@uppercase_decorator
def say_hi():
    return 'hello there'
```

```python
say_hi()
```
