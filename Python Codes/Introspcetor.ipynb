```python
class EnsureDescribe(type) :

    def __new__(cls, name, bases, attrs):
        if "describe" not in attrs:
            def describe(self):
                return f"{name} object - default description"
            attrs["describe"] = describe
        return super().__new__(cls,name,bases,attrs)

class A(metaclass=EnsureDescribe):
    pass

a=A()
print(a.describe())     #calling describe which is added dynamically during class creation
```
